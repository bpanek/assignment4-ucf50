```python
import sys
from pathlib import Path

REPO_ROOT = Path.cwd()

if not (REPO_ROOT / "run.py").is_file():
    raise RuntimeError(
        f"The kernel started in {REPO_ROOT}. "
        "Start JupyterLab from the cloned repo folder that contains run.py."
    )

sys.path.insert(0, str(REPO_ROOT))
from run import evaluate_window_lrcn, train_window_lrcn

VIDEO_ROOT = REPO_ROOT / "data" / "UCF50"
MANIFEST_PATH = REPO_ROOT / "ucf50_split_seed0.json"
OUTPUT_DIR = REPO_ROOT / "outputs"

if not MANIFEST_PATH.is_file():
    raise FileNotFoundError(f"Split manifest not found at {MANIFEST_PATH}")

if not (VIDEO_ROOT / "BaseballPitch").is_dir():
    raise FileNotFoundError(
        f"Extract UCF50.rar into {REPO_ROOT / 'data'} so the class folders "
        f"are directly inside {VIDEO_ROOT}."
    )
```

```python
checkpoint_path, loss_history, accuracy_history = train_window_lrcn(
    VIDEO_ROOT, MANIFEST_PATH, OUTPUT_DIR
)

validation_scores = accuracy_history["val"]
best_index = validation_scores.index(max(validation_scores))

print(f"Selected epoch: {best_index + 1}")
print(f"Validation accuracy: {100 * validation_scores[best_index]:.2f}%")
print(f"Validation loss: {loss_history['val'][best_index]:.4f}")
print(f"Checkpoint: {checkpoint_path}")
```

```python
test_accuracy = evaluate_window_lrcn(
    VIDEO_ROOT, MANIFEST_PATH, checkpoint_path
)

print(f"93% test accuracy requirement met: {test_accuracy >= 0.93}")
```
